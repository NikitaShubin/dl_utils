```python
import os
import sys
from pathlib import Path

os.environ['CUDA_DEVICE_ORDER'] = 'PCI_BUS_ID'
os.environ['CUDA_VISIBLE_DEVICES'] = '0'
os.environ['ONNX_MODE'] = 'cuda'
os.environ['KERAS_MODULE'] = 'tf_keras'
os.environ['TOKENIZERS_PARALLELISM'] = 'true'

# Ищем корень проекта вверх от текущей папки: так ячейка работает и из корня,
# и из вложенных директорий с ноутбуками.
cwd = Path.cwd()
project_root = next(
    (
        parent
        for parent in (cwd, *cwd.parents)
        if (parent / '3rdparty' / 'dl_utils').is_dir()
    ),
    None,
)
if project_root is None:
    msg = 'Не найден сабмодуль 3rdparty/dl_utils'
    raise RuntimeError(msg)

sys.path.insert(0, str(project_root / '3rdparty' / 'dl_utils'))
if (project_root / 'src').is_dir():
    sys.path.insert(0, str(project_root / 'src'))
```
